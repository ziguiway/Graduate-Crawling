## 基础哈希表

哈希表是一类数据结构，能够对数据建立索引，从而提升查找操作的效率。
在本节中，你将看到最简单哈希函数的实现。

```python
import numpy as np                # library for array and matrix manipulation
import pprint                     # utilities for console printing 
from utils_nb import plot_vectors # helper function to plot vectors
import matplotlib.pyplot as plt   # visualization library

pp = pprint.PrettyPrinter(indent=4) # Instantiate a pretty printer
```

在下一个代码单元中，我们将定义一个简单的整数哈希函数。该函数接收一个整数列表以及预期的桶数量。函数会生成一个以字典形式存储的哈希表：字典的键为哈希键，对应的值是输入列表中经过哈希映射后的元素。

这里使用的哈希函数计算方式为：每个元素与桶总数进行整数除法后取余数。

```python
def basic_hash_table(value_l, n_buckets):
    
    def hash_function(value, n_buckets):
        return int(value) % n_buckets
    
    hash_table = {i:[] for i in range(n_buckets)} # Initialize all the buckets in the hash table as empty lists

    for value in value_l:
        hash_value = hash_function(value,n_buckets) # Get the hash key for the given value
        hash_table[hash_value].append(value) # Add the element to the corresponding bucket
    
    return hash_table
```

接下来我们看哈希表函数的实际运行效果。美化打印函数（`pprint()`）可以输出可读性更强的结果。

```python
value_l = [100, 10, 14, 17, 97] # Set of values to hash
hash_table_example = basic_hash_table(value_l, n_buckets=10)
pp.pprint(hash_table_example)
```

在这个例子里，桶编号恰好等于每个数字的最后一位。

## 平面

多平面哈希函数是另一类哈希函数。它的核心思想是：对由 n 个平面相交划分出的每一个区域进行编号。下面的代码将展示多平面原理最基础的形式，首先从单个平面开始介绍：

```python
P = np.array([[1, 1]]) # Define a single plane. 
fig, ax1 = plt.subplots(figsize=(8, 8)) # Create a plot

plot_vectors([P], axes=[2, 2], ax=ax1) # Plot the plane P as a vector

# Plot  random points. 
for i in range(0, 10):
        v1 = np.array(np.random.uniform(-2, 2, 2)) # Get a pair of random numbers between -4 and 4 
        side_of_plane = np.sign(np.dot(P, v1.T)) 
        
        # Color the points depending on the sign of the result of np.dot(P, point.T)
        if side_of_plane == 1:
            ax1.plot([v1[0]], [v1[1]], 'bo') # Plot blue points
        else:
            ax1.plot([v1[0]], [v1[1]], 'ro') # Plot red points

plt.show()
```

首先需要注意：定义平面的向量并不代表平面两侧的分界线，它标识的是平面“正侧”所在的方向。这一点并不直观！

如果我们想要画出分隔平面，就需要绘制一条与向量 `P` 垂直的直线。我们可以借助 $90^\circ$ 旋转矩阵得到这条直线。

你可以自行修改平面向量 `P` 的方向。

```python

P = np.array([[1, 2]])  # 定义单个平面，你可以修改它的方向

# 获取与向量P垂直的平面，这里使用旋转矩阵实现
PT = np.dot([[0, 1], [-1, 0]], P.T).T  

fig, ax1 = plt.subplots(figsize=(8, 8)) # 创建自定义尺寸的画布

plot_vectors([P], colors=['b'], axes=[2, 2], ax=ax1) # 将平面P以向量形式绘制出来

# 使用两个向量画出分割平面
# 乘以2缩放只是为了让箭头延伸出当前视图范围
plot_vectors([PT * 4, PT * -4], colors=['k', 'k'], axes=[4, 4], ax=ax1)

# 绘制20个随机点
for i in range(0, 20):
        v1 = np.array(np.random.uniform(-4, 4, 2)) # 生成范围[-4,4]内的一组二维随机数
        side_of_plane = np.sign(np.dot(P, v1.T)) # 计算与P做点积并获取结果符号
        # 根据点积结果的符号对点进行着色
        if side_of_plane == 1:
            ax1.plot([v1[0]], [v1[1]], 'bo') # 绘制蓝色圆点
        else:
            ax1.plot([v1[0]], [v1[1]], 'ro') # 绘制红色圆点

plt.show()
```

接下来，我们来分析为各个点上色的代码逻辑。

```python
P = np.array([[1, 1]])      # Single plane
v1 = np.array([[1, 2]])     # Sample point 1
v2 = np.array([[-1, 1]])    # Sample point 2
v3 = np.array([[-2, -1]])   # Sample point 3
```

The function below checks in which side of the plane P is located the vector `v`

```python
def side_of_plane(P, v):
    dotproduct = np.dot(P, v.T) # Get the dot product P * v'
    sign_of_dot_product = np.sign(dotproduct) # The sign of the elements of the dotproduct matrix 
    sign_of_dot_product_scalar = sign_of_dot_product.item() # The value of the first item
    return sign_of_dot_product_scalar
```

```python
side_of_plane(P, v1) # In which side is [1, 2]
```

```python
side_of_plane(P, v2) # In which side is [-1, 1]
```

```python
side_of_plane(P, v3) # In which side is [-2, -1]
```

## Hash Function with multiple planes

In the following section, we are going to define a hash function with a list of three custom planes in 2D.

```python
P1 = np.array([[1, 1]])   # First plane 2D
P2 = np.array([[-1, 1]])  # Second plane 2D
P3 = np.array([[-1, -1]]) # Third plane 2D
P_l = [P1, P2, P3]  # List of arrays. It is the multi plane

# Vector to search
v = np.array([[2, 2]])
```

The next function creates a hash value based on a set of planes. The output value is a combination of the side of the plane where the vector is localized with respect to the collection of planes.

We can think of this list of planes as a set of basic hash functions, each of which can produce only 1 or 0 as output.

```python
def hash_multi_plane(P_l, v):
    hash_value = 0
    for i, P in enumerate(P_l):
        sign = side_of_plane(P,v)
        hash_i = 1 if sign >=0 else 0
        hash_value += 2**i * hash_i
    return hash_value
```

```python
hash_multi_plane(P_l, v) # Find the number of the plane that containes this value
```

## Random Planes

In the cell below, we create a set of three random planes

```python
np.random.seed(0)
num_dimensions = 2 # is 300 in assignment
num_planes = 3 # is 10 in assignment
random_planes_matrix = np.random.normal(
                       size=(num_planes,
                             num_dimensions))
print(random_planes_matrix)
```

```python
v = np.array([[2, 2]])
```

The next function is similar to the `side_of_plane()` function, but it evaluates more than a plane each time. The result is an array with the side of the plane of `v`, for the set of planes `P`

```python
# Side of the plane function. The result is a matrix
def side_of_plane_matrix(P, v):
    dotproduct = np.dot(P, v.T)
    sign_of_dot_product = np.sign(dotproduct) # Get a boolean value telling if the value in the cell is positive or negative
    return sign_of_dot_product
```

Get the side of the plane of the vector `[2, 2]` for the set of random planes. 

```python
sides_l = side_of_plane_matrix(
            random_planes_matrix, v)
sides_l
```

Now, let us use the former function to define our multiplane hash function

```python
def hash_multi_plane_matrix(P, v, num_planes):
    sides_matrix = side_of_plane_matrix(P, v) # Get the side of planes for P and v
    hash_value = 0
    for i in range(num_planes):
        sign = sides_matrix[i].item() # Get the value inside the matrix cell
        hash_i = 1 if sign >=0 else 0
        hash_value += 2**i * hash_i # sum 2^i * hash_i
        
    return hash_value
```

Print the bucket hash for the vector `v = [2, 2]`. 

```python
hash_multi_plane_matrix(random_planes_matrix, v, num_planes)
```

## Document vectors

Before we finish this lab, remember that you can represent a document as a vector by adding up the word vectors for the words inside the document. In this example, our embedding contains only three words, each represented by a 3D array.

```python
word_embedding = {"I": np.array([1,0,1]),
                   "love": np.array([-1,0,1]),
                   "learning": np.array([1,0,1])
                  }
words_in_document = ['I', 'love', 'learning', 'not_a_word']
document_embedding = np.array([0,0,0])
for word in words_in_document:
    document_embedding += word_embedding.get(word,0)
    
print(document_embedding)
```
