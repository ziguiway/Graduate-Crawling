### 导入模块与准备数据

```python
# data
word = 'dearz' # 🦌
```

### Splits
Find all the ways you can split a word into 2 parts !

```python
# splits with a loop
splits_a = []
for i in range(len(word)+1):
    splits_a.append([word[:i],word[i:]])

for i in splits_a:
    print(i)
```

```python
# same splits, done using a list comprehension
splits_b = [(word[:i], word[i:]) for i in range(len(word) + 1)]

for i in splits_b:
    print(i)
```

### 删除编辑操作
对`splits`列表内每组字符串执行删除字母操作。
<br>
该操作的作用：遍历原单词所有位置，依次删掉每一个可能的字符。

```python
# deletes with a loop
splits = splits_a
deletes = []

print('word : ', word)
for L,R in splits:
    if R:
        print(L + R[1:], ' <-- delete ', R[0])
```

我们值得仔细观察“删除”操作的执行过程。
<br>
以`splits`列表里的第一项为例：

```python
# 拆解过程
print('原单词：', word)
one_split = splits[0]
print('splits列表中的第一项：', one_split)
L = one_split[0]
R = one_split[1]
print('L：', L)
print('R：', R)
print('*** 通过舍弃R的首字符，等价完成删除操作 ***')
print('L + R[1:]：',L + R[1:], ' <-- 被删除字符：', R[0])
```

因此最终效果就是将 **'dearz'** 删除第一个字符，得到 **'earz'**。
<br>
你可以使用上方代码块里的**循环**，或是下方代码块的**列表推导式**
<br>
对整个`splits`列表批量执行该操作。

```python
# deletes with a list comprehension
splits = splits_a
deletes = [L + R[1:] for L, R in splits if R]

print(deletes)
print('*** which is the same as ***')
for i in deletes:
    print(i)
```
